A while loop repeats the code **as long as a condition is True**. 

Inshort while loop means, keep doing this task while this condition is true.

```python
# Basic while loop

number = 19

while number <= 5:
    print(number)
    number += 1
```


**Every well-behaved while loop has three parts:**
1. **Initialization** — set up the variable before the loop
2. **Condition** — the test at the top
3. **Update** — change the variable inside the body so the condition eventually fails

Miss any of the three and you have a bug.

```python
count = 1              # 1. initialization
while count <= 15:      # 2. condition
    print(count)
    count += 1         # 3. update

print("Loop finished. count =", count)
```

loop variable should usually be updated inside a while loop.

Without updating the variable, the condition may never become False

The loop variable must be updated so the while loop can stop.

There are some common mistakes:
1. **No update** → Infinite loop.
2. **Wrong update position** → Can change the output.
3. **Off-by-one** → < and <= give different results.
4. **Update inside if** → The loop may get stuck.

```python
number = 1

while number <= 12:
    print(number)
    number += 2
```

```python
number = 10

while number >= 0:
    print(number)
    number -= 2
```

```python
# false condition from the start

n = 10
while n < 5:
    print("this never prints")
print("body ran 0 times")
```

```python
items = [1, 2, 3]
while items:
    print("popped:", items.pop())
print("list is now empty:", items)
```

```python
n = 3
while n:
    print("n =", n)
    n -= 1
```

```python
#update placement changes the output

print("update AFTER the work:")
i = 1
while i <= 3:
    print(" ", i)     # 1, 2, 3
    i += 1

print("update BEFORE the work:")
i = 1
while i <= 3:
    i += 1
    print(" ", i)     # 2, 3, 4
```

## Infinite Loops

loop that never stops because its condition always remains True.

To avoid infinite loops, make sure the loop condition can eventually become False.

```python
count = 12

while count <= 25:
    print(count)
    count += 1
```

```python
number = 10

while number > 0:
    print(number)
    number -= 1
```

```python
x = 0.0
for _ in range(10):
    x += 0.1
print("0.1 added ten times =", x)
print("Is it exactly 1.0?", x == 1.0)
```

## break

break exits the loop **immediately**, the rest of the body is skipped and the condition is not rechecked.

It is useful when we want to stop the loop before the condition becomes False.

break only exits the **innermost** loop it sits in. In nested loops, an inner break returns control to the outer loop; it does not escape both.

example: A program keeps checking passwords, but stops immediately when the correct password is entered.

```python
correct_password = "python123"

while True:
    password = input("Enter password: ")

    if password == correct_password:
        print("Access granted!")
        break

    print("Wrong password!")
```

## continue

continue skips the rest of the current iteration and moves to the next iteration.

Difference:

break → completely stops the loop.

continue → skips the current iteration but keeps the loop running.

Example: We want to print numbers from 1 to 5 but skip number 3.

```python
number = 0

while number < 5:
    number += 1

    if number == 3:
        continue

    print(number)
```

```python
# break

number = 1

while number <= 10:
    if number == 5:
        break

    print(number)
    number += 1
```

```python
# continue

number = 0

while number < 10:
    number += 1

    if number == 5:
        continue

    print(number)
```

## while...else

Python allows an else block with a while loop.

The else block runs when the loop finishes normally.

If the loop ends because of break, the else block does NOT execute.

So:

Normal completion → else runs

break → else does not run

```python
number = 1

while number <= 5:
    print(number)
    number += 1
else:
    print("Loop finished normally!")
```

```python
number = 1

while number <= 5:
    print(number)

    if number == 3:
        break

    number += 1
else:
    print("Loop finished normally!")
```

## User Input with while

while loop is very useful with input() when we want to repeatedly ask the user for information.

```python
age = int(input("Enter your age: "))

while age < 0:
    print("Age cannot be negative.")
    age = int(input("Enter your age again: "))

print("Your age is:", age)
```

```python
correct_password = "hello123"
password = input("Enter password: ")

while password != correct_password:
    print("Incorrect password!")
    password = input("Try again: ")

print("Login successful!")
```

## Accumulator Pattern

An accumulator is a variable that keeps collecting or building a result during a loop.

example: adding numbers, calculating items etc

```python
total = 0
number = 10

while number <= 25:
    total += number
    number += 1

print("Total:", total)
```

```python
total = 0
number = 1

while number <= 5:
    total += number
    number += 1

print("Total:", total)
```

```python
number = 1
count = 0

while number <= 10:

    if number % 2 == 0:
        count += 1

    number += 1

print("Number of even numbers:", count)
```

### Accumulator + User Input

We can repeatedly take numbers from the user and add them together.

A special value can be used to stop the loop.

The accumulator stores the total.

```python
total = 0

number = int(input("Enter a number (0 to stop): "))

while number != 0:
    total += number
    number = int(input("Enter a number (0 to stop): "))

print("Total:", total)
```

```python
#string accumulator

result = ""
number = 1

while number <= 5:
    result += str(number)
    number += 1

print(result)
```
