## Scope

Scope means where a variable can be accessed.

- Local variable : created inside a function. It can be used inside the function but normally cannot be accessed outside it.
- Global variable : created outside a function. 

A local variable normally cannot be accessed outside its function.
A function can read a global variable.

```python
# global variable

name = "Alex"

def greet():
    print(name)

greet()
```

```python
score = 100

def show_score():
    print(score)

show_score()
```

```python
country = "USA"

def show_country():
    print(country)

show_country()
print(country)
```

```python
#local variable

def greet():
    message = "Hello"
    print(message)

greet()
```

```python
def calculate():
    x = 90
    y = 80
    result = x + y
    
    print(result)

calculate()
```

### changing a global variable 

Reading a global variable inside a function is allowed.

But assigning to a variable inside a function normally creates a local variable.

Use the global keyword when you intentionally want to modify the global variable.

```python
score = 100

def change_score():
    score = 200
    print("Inside:", score)

change_score()

print("Outside:", score)
```

## global Keyword

Tells Python that a variable belongs to the global scope. It allows a function to modify the global variable.

```python
count = 0

def increase():
    global count
    count = count + 1

increase()

print(count)
```

```python
balance = 7000

def deposit():
    global balance
    balance += 700

deposit()

print(balance)
```

```python
count = 0

def increase():
    global count
    count += 1

increase()
increase()
increase()

print(count)
```

global works, but unnecessary global variables can make programs harder to understand.

```python
# better approach

def deposit(balance, amount):
    return balance + amount

balance = 1000
balance = deposit(balance, 500)
print(balance)
```

```python
def increase(number):
    return number + 1

x = 11
x = increase(x)
print(x)
```

## Shadowing

Shadowing happens when a local variable has the same name as a global variable.

```python
name = "Alex"

def show():
    name = "Ali"
    print(name)

show()
print(name)
```

```python
message = "Global message"

def test():
    message = "Local message"
    print(message)

test()
print(message)
```

## *args and **kwargs

*args collects extra positional arguments. The collected arguments are stored in a tuple.

**kwargs collects extra keyword arguments. The collected arguments are stored in a dictionary. 

```python
def show(*args):
    print(args)

show(10, 20, 30)
```

```python
def show(*args):
    print(args)
    print(type(args))
show(10, 20, 30)
```

```python
def show_numbers(*args):
    for number in args:
        print(number)

show_numbers(60, 80, 85, 90)
```

```python
# zero argument

def show(*args):
    print(args)

show()
```

```python
# **kwargs

def show(**kwargs):
    print(kwargs)

show(name="Siri", age=20, city="Islamabad")
```

```python
def show(**kwargs):
    print(type(kwargs))

show(name="Siri", age=20)
```

```python
def show_info(**kwargs):
    for key, value in kwargs.items():
        print(key, "=", value)

show_info(
    name="Alex",
    age=20,
    country="Pakistan"
)
```

```python
# *args + **kwargs

def test(*args, **kwargs):
    print("args:", args)
    print("kwargs:", kwargs)

test(
    10,
    20,
    30,
    name="John",
    age=20
)
```

When combining different parameter types, the order matters.

```python
def student(name, age=18, *subjects, **details):
    print("Name:", name)  #normal positional parameter
    print("Age:", age)    #default parameters
    print("Subjects:", subjects)   #*args
    print("Details:", details)   #**kwargs

student(
    "Ali",
    20,
    "Python",
    "Networking",
    city="London"
)
```

Imp : A non-default parameter cannot come after a default parameter.

### Mutable Default Argument 

Be careful when using mutable objects such as:

- list
- dictionary
- set

as default parameter values.

```python
def add_item(item, items=[]):
    items.append(item)
    return items

print(add_item("apple"))
print(add_item("banana"))
print(add_item("mango"))
```

```python
def add_number(number, numbers=[]):
    numbers.append(number)
    return numbers

print(add_number(1))
print(add_number(2))
print(add_number(3))
```

The default list is created once when the function is defined. It is not recreated every time the function is called.

Imp : Use None as the default value. Then create a new list inside the function.

```python
def add_number(number, numbers=None):
    if numbers is None:
        numbers = []

    numbers.append(number)

    return numbers

print(add_number(1))
print(add_number(2))
print(add_number(3))
```

## lambda

A lambda is a small anonymous function. lambda automatically returns the result of its expression.

```python
def square(x):
    return x * x

print(square(5))
```

```python
square = lambda x: x * x
print(square(5))
```

```python
# multiple parameters
add = lambda a, b: a + b
print(add(70, 90))
```

```python
check = lambda x: "Even" if x % 2 == 0 else "Odd"

print(check(10))
print(check(7))
```

## map()

map() applies a function to every item in an iterable.

```python
numbers = [1, 2, 3, 4, 5]

squares = map(lambda x: x * x, numbers)
print(list(squares))
```

```python
names = ["siri", "ali", "sara"]

result = map(str.upper, names)
print(list(result))
```

## filter()

filter() keeps only the items for which a condition returns True.

```python
numbers = [5, 12, 3, 20, 15, 7]

result = filter(lambda x: x > 10, numbers)
print(list(result))
```

```python
def is_even(x):
    return x % 2 == 0

numbers = [1, 2, 3, 4, 5, 6, 11, 14, 18]

result = filter(is_even, numbers)
print(list(result))
```

## sorted(key=...)

sorted() creates a new sorted list.

The key parameter tells Python what value should be used for sorting.

```python
names = ["Ali", "Muhammad", "Siri", "John"]

result = sorted(names, key=lambda name: len(name))
print(result)
```

```python
numbers = [23, 15, 42, 31, 18]

result = sorted(numbers, key=lambda x: x % 10)
print(result)
```

## Recursion

Recursion means a function calls itself.

A recursive function needs two important parts:

1. Base case : ells the recursive function when to stop.
2. Recursive case : calls the function again 

```python
def countdown(n):
    if n == 0:
        return

    print(n)
    countdown(n - 1)

countdown(5)
```

```python
# basic

def countdown(n):
    if n == 0:
        return

    print(n)
    countdown(n - 1)

countdown(3)
```

```python
# recursive
def countdown(n):
    if n == 0:
        return

    print(n)

    countdown(n - 1)

countdown(7)
```

If a recursive function has no stopping condition, it keeps calling itself.

Eventually Python stops the program and raises: RecursionError

```python
def count(n):
    if n == 0:
        return

    print(n)
    count(n + 1)

count(5)
```
