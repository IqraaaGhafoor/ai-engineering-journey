# Variables and Types
Variables are the symbolic names that references or print an object in memory.  They are used to store and manipulate data in a program.
These are basically like a container where you put data.

Think of a variable as a labeled box where you can place data and retrieve whhen needed. 
straightforward in python 
assign a value to a variable name using the assignment operator (=)
# Rules for variable naming 
1. Case sensitive
2. Start with a letter or underscore
   
 valid: _value, data1

 invalid: 1data, #score

1. Followed by leters , digits, or underscore 
   
 valid: value_1, data_set
 
 invalid: data-set, value-1 (hyphen are not allowed.)

# Variables naming conventions
1. Use lowercase letters
2. Avoid reserved keywords
3. Meaningful names

## Integer(int)

```python
age = 10 
print(type(age))
```

```python
# Basic arithmetic operations with integers
a = 555
b = 99

sum_result = a + b
difference = a - b
product = a * b
quotient = a // b  # Integer division (no decimal part)
remainder = a % b  # Modulus (remainder of division)
power = a ** 2
power = b ** 3

print("Sum:", sum_result)
print("Difference:", difference)
print("Product:", product)
print("Quotient:", quotient)
print("Remainder:", remainder)
print("Power:", power)
print("Power:", power)

# Type
print(type(sum_result))
print(type(difference))
print(type(product))
print(type(quotient))
print(type(remainder))
print(type(power))
```

```python
numbers = [33, 89, 111, 44,107]

max_value = max(numbers)
min_value = min(numbers)

print("Maximum value:", max_value)
print("Minimum value:", min_value)

print(type(max_value))
print(type(min_value))
```
