## What is module?

- A module is a Python file (.py) containing reusable code.
- contain functions, variables, classes, etc.

## import module
- import module imports the complete module.
- Access its functions or variables using module.name.
- This makes it clear where the function came from.

```python
import math

print(math.sqrt(25))
print(math.pi)
print(math.factorial(5))
```

```python
import random

print(random.randint(1, 10))
print(random.random())
```

## from module import name

- Imports a specific function or object from a module. module name is not required when using the imported name.

```python
from math import sqrt

print(sqrt(36))
print(sqrt(49))
print(sqrt(81))
```

```python
from math import sqrt, pi, factorial

print(sqrt(49))
print(pi)
print(factorial(5))
```

## import module as alias

- `as` gives a module a shorter name.
- Useful when a module name is long or commonly given an alias.

```python
import math as m

print(m.sqrt(64))
print(m.pi)
```

## Why from module import * is Discouraged

- `from module import *` imports many names at once.
- It can make code difficult to understand.
- It can cause name conflicts.
- It becomes unclear where a function or variable came from.
- Prefer import module or specific imports.

```python
# This works, but is discouraged

from math import *

print(sqrt(36))
print(pi)
```

```python
# it's better and clearer

import math
print(math.sqrt(25))
print(math.pi)
```

```python
import csv

with open("students.csv", "r") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
```

## How Python Finds Modules

- When Python sees `import math` or any other, it searches locations in its module search path.
- The project/current directory is an important location.
- `sys.path` shows the locations Python searches.
- If Python cannot find the module, it raises `ModuleNotFoundError`.

## if __name__ == "__main__":

- Every Python module has a special `__name__` variable.
- When a file is run directly, `__name__` becomes `"__main__"`.
- When the file is imported, `__name__` becomes the module's name.
- The main guard makes code run only when the file is executed directly.

```python
def add(a, b):
    return a + b


if __name__ == "__main__":
    print(add(10, 20))
```

The main guard prevents testing/program code from running automatically when the module is imported.


```python
def greet(name):
    return f"Hello, {name}!"


if __name__ == "__main__":
    print(greet("Ali"))
```

## Packages

- A package is a folder used to organize related Python modules.
- A package can contain multiple `.py` files.
- `__init__.py` is commonly used inside the package.

## __init__.py

- `__init__.py` is a special file commonly placed inside Python packages.
- It can contain package initialization code.
- It also helps clearly identify a regular Python package.

## 26. Standard Library vs Third-Party Packages

Standard library:
- Comes with Python.
- Usually does not require `pip install`.
- eg: `math`, `random`, `json`, `os`.

Third-party package:
- Created outside Python's standard library.
- Usually installed separately.
- eg: `numpy`, `pandas`.

## pip

- `pip` is Python's package installer.
- It is used to install third-party packages.
- Install:
  `pip install package_name`
- Use the package in Python with `import`.

## Virtual Environments

- A virtual environment is an isolated Python environment for a project.
- Each project can have its own packages and package versions.
- It helps prevent dependency conflicts.
- Miniconda also provides environment management.

## requirements.txt

- Stores the third-party packages required by a project.
- Helps others recreate the project's environment.
- Install packages from the file using:
  `pip install -r requirements.txt`
