### Accessing Dictionary Values

Two common ways:

- d[key] → gives the value, but raises KeyError if the key is missing.
- d.get(key) → safely gets the value and returns None if the key is missing.
- d.get(key, default) → returns a default value when the key is missing.

```python
student = {
    "name": "Meddy",
    "age": 20,
    "city": "America"
}

print(student["name"])
print(student["age"])
print(student["city"])
```

```python
# Using get()

print(student.get("name"))
print(student.get("age"))
print(student.get("phone"))
```

```python
# get() with a default value

print(student.get("phone", "Not available"))
print(student.get("country", "Not provided"))
```

### Adding and Updating Dictionary Items

d[key] = value does two jobs:

- If the key does not exist → adds a new item.
- If the key already exists → updates its value.

update() can add or update multiple items at once.

```python
student = {
    "name": "Alex",
    "age": 25
}

# Adding a new key

student["city"] = "London"

print(student)
```

```python
# Updating an existing key

student["age"] = 26
print(student)
```

```python
# adding and updating
student.update({
    "age": 22,
    "country": "UK",
    "course": "Python"
})

print(student)
```

### Removing Dictionary Items

- pop(key) : removes a specific key and returns its value.
- popitem() : removes and returns the last inserted key-value pair.
- del d[key] : deletes a specific key.
- clear() : removes all items.

```python
student = {
    "name": "zarwa",
    "age": 20,
    "city": "Islamabad"
}

removed = student.pop("age")

print("Removed:", removed)
print(student)
```

```python
# popitem()

student = {
    "name": "Zarwa",
    "age": 20,
    "city": "Islamabad"
}

item = student.popitem()

print("Removed:", item)
print(student)
```

```python
# del

student = {
    "name": "Ali",
    "age": 25,
    "city": "Islamabad"
}

del student["age"]
print(student)
```

```python
# clear()

student = {
    "name": "Ali",
    "age": 20,
    "city": "Islamabad"
}

student.clear()
print(student)
```

For dictionaries, in checks **keys**, not values.

```python
student = {
    "name": "Ali",
    "age": 20,
    "city": "Islamabad"
}

print("name" in student)
print("age" in student)
print("phone" in student)
```

```python
# Checking values

print("Ali" in student.values())
print(20 in student.values())
print("Pakistan" in student.values())
```

```python
if "name" in student:
    print("Name exists")
else:
    print("Name does not exist")
```

Three important methods:

- keys() → returns dictionary keys
- values() → returns dictionary values
- items() → returns key-value pairs

These are commonly used with loops.

```python
student = {
    "name": "Ali",
    "age": 20,
    "city": "Islamabad"
}

print(student.keys())
print(student.values())
print(student.items())
```

```python
# loop through keys

for key in student.keys():
    print(key)
```

```python
# loop through items

for item in student.items():
    print(item)
```

```python
# loop through values

for value in student.values():
    print(value)
```

```python
student = {
    "name": "Alex",
    "age": 20,
    "city": "Islamabad"
}

for key, value in student.items():
    print(key, "=", value)
```

```python
marks = {
    "Math": 90,
    "English": 85,
    "Python": 95
}

for subject, marks_obtained in marks.items():
    print(subject, ":", marks_obtained)
```

```python
marks = {
    "Math": 70,
    "English": 87,
    "Python": 95
}

total = 0

for subject, marks_obtained in marks.items():
    total += marks_obtained

print("Total:", total)
```

### Nested Dictionaries

A dictionary can contain another dictionary.

This is useful for storing structured information such as multiple students, users, employees, or products.

```python
students = {
    "student1": {
        "name": "Ali",
        "age": 20,
        "marks": 85
    },
    "student2": {
        "name": "Sara",
        "age": 21,
        "marks": 92
    }
}

print(students)
```

```python
print(students["student1"]["name"])
print(students["student1"]["marks"])
print(students["student2"]["name"])
print(students["student2"]["marks"])
```

```python
students["student1"]["marks"] = 90
print(students["student1"])
```

```python
# loop through nested dictionary

for student_id, info in students.items():
    print("ID:", student_id)
    print("Name:", info["name"])
    print("Age:", info["age"])
    print("Marks:", info["marks"])
    print()
```

### setdefault()

setdefault() gets a value if the key exists.

If the key does not exist, it creates the key with the given default value.

It does not replace an existing value.

```python
student = {
    "name": "Iqra"
}

age = student.setdefault("age", 20)
print(age)
print(student)
```

```python
students = {}

students.setdefault("Python", []).append("Ali")
students.setdefault("Python", []).append("Sara")
students.setdefault("Cyber Security", []).append("Ahmed")
print(students)
```

Dictionary comprehension provides a short way to create dictionaries.

```python
# Normal

squares = {}

for number in range(1, 6):
    squares[number] = number ** 2

print(squares)
```

```python
squares = {
    number: number ** 2
    for number in range(1, 6)
}

print(squares)
```

```python
even_squares = {
    number: number ** 2
    for number in range(1, 11)
    if number % 2 == 0
}

print(even_squares)
```

```python
# Create a dictionary of cubes

cubes = {
    number: number ** 3
    for number in range(1, 11)
}

print(cubes)
```
