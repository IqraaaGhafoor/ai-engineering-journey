### open() and File Modes

open() is used to open a file.

Basic syntax:
open(filename, mode)

There are some important modes:

- r : read
- w : write; erases old content
- a : append; keeps old content
- r+ : read and write
- rb : read binary
- wb : write binary
- ab : append binary

Remember:
- w can silently erase existing content.
- a adds new content at the end.

```python
# r - read

file = open("notes.txt", "r")
print(file.read())
file.close()
```

```python
# w - write

file = open("notes.txt", "w")
file.write("New content")
file.close()
```

```python
# a - append

file = open("notes.txt", "a")
file.write("\nMore content")
file.close()
```

```python
# r+ - read and write

file = open("notes.txt", "r+")
print(file.read())
file.write("\nPython")
file.close()
```

```python
# Binary modes

file = open("data.bin", "wb")
file.write(b"Hello")
file.close()
```

## w vs a

- w : removes old content and writes new content. you have to use w carefully because existing file content can be lost
- a : keeps old content and adds new content at the end.

```python
# w removes previous content

with open("test.txt", "w") as f:
    f.write("First text")
```

```python
# Running w again replaces the content

with open("test.txt", "w") as f:
    f.write("New text")
```

```python
# a keeps existing content

with open("test.txt", "a") as f:
    f.write("\nAdded text")
```

## read()
- read() reads the contents of the file.
- Without an argument : reads the entire file.
- read(n) : reads up to n characters.

## readline()
- readline() reads one line at a time.
- Calling it again reads the next line.

## readlines()
- readlines() reads all lines.
- It returns the lines as a list.
- Newline characters \n may be included.

## Reading Methods
- read() : entire file as one string
- readline() : one line
- readlines() : all lines as a list
- for line in f : processes the file line by line

```python
# read()

with open("notes.txt", "r") as f:
    print(f.read())
```

```python
# readline()

with open("notes.txt", "r") as f:
    print(f.readline())
```

```python
# readlines()

with open("notes.txt", "r") as f:
    print(f.readlines())
```

## Looping Through a File

A file object can be used directly in a for loop.
- Reads one line at a time.
- More memory-friendly for large files.
- Avoids loading the entire file into memory at once.

```python
with open("notes.txt", "r") as f:
    for line in f:
        print(line)
```

```python
# Better when processing each line

with open("notes.txt", "r") as f:
    for line in f:
        print(line.strip())
```

## .strip()
- .strip() removes whitespace from the beginning and end of a string.
- When reading files, it is commonly used to remove the newline \n.

## write()
- write() writes a string into a file.
- It does not automatically add a newline.
- Use \n to move to the next line.

## writelines()
- writelines() writes multiple strings.
- It does not automatically add \n.
- Add \n yourself when separate lines are needed.

## close()

- close() closes an opened file.
- Closing releases file resources.
- It also helps ensure written data is properly handled.

IMP: When using open() directly, remember to close the file.

## with open(...) as f
with creates a context manager for file handling.

- Automatically closes the file.
- Safer and cleaner than manually using close().
- This should be the preferred method from now on.

```python
with open("notes.txt", "r") as f:
    for line in f:
        line = line.strip()
        print(line)
```

```python
with open("subjects.txt", "w") as f:
    f.write("Python")
```

```python
lines = [
    "Python\n",
    "Networking\n",
    "Cybersecurity\n",
    "Artificial Intelligence\n"
]

with open("subjects.txt", "w") as f:
    f.writelines(lines)
```

```python
file = open("notes.txt", "r")

data = file.read()
print(data)
file.close()
```

```python
with open("notes.txt", "r") as f:
    data = f.read()

print(data)
```

```python
# Writing with with

with open("notes.txt", "w") as f:
    f.write("Hello Python")
```

## FileNotFoundError

FileNotFoundError occurs when Python cannot find the file or path.

Common causes:
- File does not exist.
- Wrong filename.
- Wrong folder.
- Incorrect file path.

Use `try` and `except` to handle it.

## File Paths

A file path tells Python where a file is located.

### Relative path
Path relative to the current working directory.

Example:
`"data/notes.txt"`

### Absolute path
Complete location of the file.

Example:
`r"C:\Users\Name\Documents\notes.txt"`

Python uses the current working directory (cwd) when a relative path is given.

CWD is folder where python is currently working.

```python
import os
print(os.getcwd())
```

### CSV Files

CSV = Comma-Separated Values

CSV files store table-like data.

A simple CSV can be read using split(",").
For real CSV files, the csv module is safer and more reliable.

### Reading CSV with split(",")

split(",") separates a line wherever a comma appears.

Useful for understanding simple CSV files, but it is not reliable for complex CSV data containing quoted commas.
