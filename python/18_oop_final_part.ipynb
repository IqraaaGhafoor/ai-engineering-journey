## `__str__()`

- `__str__()` controls what `print(object)` displays.
- It should return a string.
- It makes objects easier to read.
- Without it, Python usually shows a less useful object representation.

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Student: {self.name}, Age: {self.age}"

student1 = Student("Ali", 20)
print(student1)
```

### `__str__()` With more info

- We can include multiple attributes.
- `__str__()` should return a readable description of the object.

```python
class Product:

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name} costs ${self.price}"

product1 = Product("Laptop", 1000)
print(product1)
```

```python
class Fruit:

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name} costs ${self.price}"

Fruit1 = Fruit("Mnago", 10)
print(Fruit1)
```

## Inheritance

- Inheritance allows a child class to reuse code from a parent class.
- Parent class : general/common behavior.
- Child class : inherits and can add its own behavior.
- It helps avoid repeating code.

```python
class Fruit:     #here fruit is parent class

    def eat(self):
        print("Eating fruit are better for health")


class Apple(Fruit):    #apple inherits from fruit
    pass


apple1 = Apple()
apple1.eat()
```

Inheritence is useful when classes have a relationship.

Child classes can reuse it instead of writing it again.

```python
class Fruit:    #parent class

    def eat(self):
        print("Eating fruit")

    def wash(self):
        print("Washing fruit")


class Apple(Fruit):     #child class

    def peel(self):
        print("Peeling apple")


class Mango(Fruit):    #child class

    def cut(self):
        print("Cutting mango")


apple = Apple()
mango = Mango()

apple.eat()
apple.peel()

mango.eat()
mango.cut()
```

## Inheritance With `__init__()`

- A child class can have its own `__init__()`.
- Sometimes we also need the parent's initialization.
- `super()` gives access to the parent class.
- `super().__init__()` calls the parent's `__init__()`

```python
class Vehicle:

    def __init__(self, brand):
        self.brand = brand


class Car(Vehicle):

    def __init__(self, brand, model):
        super().__init__(brand)
        self.model = model


car1 = Car("BMW", "Rools Ryce")

print(car1.brand)
print(car1.model)
```

### Understanding `super().__init__()`

**Parent:**

`Vehicle.__init__()` creates `self.brand`.

**Child:**

`Car.__init__()` creates `self.model`.

`super().__init__(brand)` tells Python:

"Run the parent's `__init__()` using this object."

This allows the child to reuse the parent's initialization code.

```python
class Person:

    def __init__(self, name):
        self.name = name


class Student(Person):

    def __init__(self, name, course):
        super().__init__(name)
        self.course = course


student1 = Student("Ali", "Cyber Security")

print(student1.name)
print(student1.course)
```

### Method Overriding

- Method overriding happens when a child class defines a method with the same name as the parent.
- The child's version is used for the child object.
- This allows the child to provide its own behavior.

```python
class Animal:

    def sound(self):
        print("Some animal sound")


class Dog(Animal):

    def sound(self):
        print("Woof!")


animal = Animal()
dog = Dog()
animal.sound()
dog.sound()
```

### Overriding + `super()`

- A child can override a parent's method.
- `super().method()` can still call the parent's version.
- This is useful when we want to keep the parent's behavior and add something extra.

```python
class Animal:

    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):

    def sound(self):
        super().sound()
        print("Dog says Woof!")

dog = Dog()
dog.sound()
```

## Encapsulation

Encapsulation means controlling how data is accessed or modified.

Python mainly uses naming conventions:

- `name` : normal/public attribute
- `_name` : internal/protected by convention
- `__name` : name mangling

Imp:
Python does not enforce private/protected access in the same way as some other languages.

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self._age = age


student1 = Student("Ali", 25)

print(student1.name)
print(student1._age)
```

## `_name`  (single underscore)

- `_name` means "this is intended for internal use."
- It is mainly a convention.
- Python still allows you to access it directly.
- It does NOT make the attribute truly private.

```python
class Student:

    def __init__(self, name):
        self._name = name

student1 = Student("Ali")
print(student1._name)
```

## `__name` (double underscore)

- `__name` triggers **name mangling**.
- Python changes the internal attribute name.
- This helps prevent accidental access or accidental overriding.
- It is not absolute security/privacy.

```python
class Student:

    def __init__(self, name):
        self.__name = name

student1 = Student("Ali")   
print(student1._Student__name)                
```

## Name Mangling

If a class contains:

`self.__name`

Python internally changes it approximately to:

`self._ClassName__name`

This process is called **name mangling**.

Purpose:
- Reduce accidental access.
- Avoid naming conflicts in inheritance.

It does not mean the data is completely private.

```python
class Student:

    def __init__(self, name):
        self.__name = name

student1 = Student("Ali")
print(student1._Student__name)
```

## `@property`

- `@property` lets us use a method like an attribute.
- Useful for computed or controlled attributes.
- We can calculate a value when it is accessed.
- It allows cleaner syntax.

Without property:
e.g `student.get_age()`

With property:
e.g `student.age`

A property is useful when a value is calculated from other attributes.

`@property` allows us to access the calculated value like an attribute.

```python
class Rectangle:

    def __init__(self, length, width):
        self.length = length
        self.width = width

    @property
    def area(self):
        return self.length * self.width

rectangle = Rectangle(10, 5)
print(rectangle.area)
```

```python
class Student:

    def __init__(self, marks):
        self.marks = marks

    @property
    def percentage(self):
        return self.marks / 500 * 100

student = Student(450)
print(student.percentage)
```
