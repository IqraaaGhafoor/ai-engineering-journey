## Array Operations

There are two types of operations:
- Scalar operation
- Vector operation

### Scalar Operation
In scalar operation, we can operate with single number. e.g a + 5 here 5 is scalar.

### Vector Operation 
In vector opertion, we can apply any operation on two numpy arrays. 
e.g a1 + a2

```python
a = np.array([1, 2, 3, 4, 5])
print(a + 5)
```

```python
a = np.array([11, 14, 16, 18])
print(a ** 3)
```

```python
a = np.array([11, 14, 16, 18])
print(a * 3)
```

```python
a1 = np.array([12, 14, 16, 18])
a2 = np.array([10, 20, 30, 40])
print(a1 + a2)
print(a1 * a2)
print(a1 ** a2)
print(a1 - a2)
```

## mean() and median()

### mean() 

Means is the average

**formula: sum of value / no of value**

### median()

Middle value after the data has been arraanged in the sorted order.

If there is an even number of values, the median is the average of the two middle values.

```python
numbers = ([10, 20, 30, 40, 50, 60])
print(np.mean(numbers))
print(np.median(numbers))
```

```python
numbers = ([11, 23, 30, 47, 50, 60])
print(np.mean(numbers))
print(np.median(numbers))
```

```python
a = np.array([[11, 13, 17], [21, 24, 28]])
print(np.mean(a))
print(np.median(a))
print(np.mean(a, axis=0))
print(np.median(a, axis=0))
```

```python
print(np.mean(a, axis=1))
print(np.median(a, axis=1))
```

## Standard Deviation

Tells us about the spread of values around the mean.
- Small standard deviation : values are closer together.
- Large standard deviation : values are more spread out.

```python
a = np.array([10, 13, 18, 20])
print(np.std(a))
```

```python
a = np.array([[10, 13, 18, 20], [21, 17, 19, 12]])
print(np.std(a))
```

```python
print(np.std(a, axis=0))
print(np.std(a, axis=1))
```

## Variance 

Variance is another measure of how spread out data is around its mean.

```python
numbers = np.array([10, 10, 10, 10])
print(np.var(numbers))
```

```python
numbers = np.array([1, 2, 3, 4, 5])
print(np.var(numbers))
```

```python
# overall operations
data = np.array([12, 18, 25, 30, 15, 20, 35, 55])

print("Maximum:", np.max(data))
print("Minimum:", np.min(data))
print("Sum:", np.sum(data))
print("Product:", np.prod(data))
print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Standard deviation:", np.std(data))
print("Variance:", np.var(data))
```

## `axis` in NumPy

When working with a 2D array, functions such as `sum()`, `mean()`, `max()`, and `min()` can operate along an axis.

For a 2D array:

- `axis=0` : operate down the rows and get a result for each column.
- `axis=1` : operate across the columns and get a result for each row.

This is very important when working with datasets.

```python
marks = np.array([
    [80, 90, 70],
    [60, 75, 85]
])

print(marks)
print(np.sum(marks, axis=1))
print(np.sum(marks, axis=0))
print(np.min(marks, axis=1))
print(np.min(marks, axis=1))
print(np.max(marks, axis=1))
print(np.max(marks, axis=1))
```
