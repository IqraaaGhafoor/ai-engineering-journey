# Sigmoid Function

Mathematical function that convert any number in to a value between 0 and 1.
It used in Machine Learning(ML), especially for binary classification.

Formula:

σ(x) = 1 / (1 + e⁻ˣ)

```python
x = np.array([10, 30, 50, 70, 90])
print(1/(1 + np.exp(-x)))     #sigmoid function
```

We can apply the sigmoid function to an entire array at once.

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
x = np.array([200, 300, 400, 500])
result = sigmoid(x)
print(result)
```

```python
x = np.array([-3, -2, -1, 0, 1, 2, 3])
result = sigmoid(x)
print(result)
```

```python
# 2D array

x = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

result = sigmoid(x)
print(result)
```

# Mean Square Error

Method used to measure how predicted values are far from the actual values.

Formula:

MSE = (1/n) × Σ(actual - predicted)²

Where:

- `actual` = actual values
- `predicted` = predicted values
- `n` = number of values

The errors are squared before taking their average.

```python
actual = np.array([10, 20, 30])
predicted = np.array([12, 18, 29])

mse = np.mean((actual - predicted)**2)

print(mse)
```

```python
actual = np.array([122, 324, 520])
predicted = np.array([126, 328, 530])

mse = np.mean((actual - predicted)**2)

print(mse)
```

Imp:

- Smaller MSE means predictions are closer to the actual values.
- Larger MSE means predictions are farther away from actual values

```python
def mse(actual, predicted):
    return np.mean((actual - predicted) ** 2)
```

```python
actual = np.array([10, 20, 30])
predicted = np.array([12, 19, 28])

print(mse(actual, predicted))
```

```python
# 2D 
actual = np.array([
    [10, 20, 30]<
    [40, 50, 60]
])

predicted = np.array([
    [70, 80, 90],
    [100, 110, 120]
])

print(mse(actual, predicted))
```

# Missing Values

Numpy represent missing values using NaN, means not a number.

`np.nan`

```python
a = np.array([10, 20, 30, np.nan, np.nan, 40, 50])
print(a[0])
print([1])
print(a[2])
print(a[3])
print(a[4])
print(a[5])
```

## `np.isnan()`

We can also detect nan values in numpy by using `np.isnan()`. It checks every element and returns True and False.
Here:
- True means Value is Nan 
- Falso means Value is not Nan

```python
a = np.array([10, 20, 30, np.nan, np.nan, 40, 50])
result = np.isnan(a)
print(result)
```

We can also count missing values. Use `np.sum()` to count the True values

```python
a = np.array([10, 20, 30, np.nan, np.nan, 40, 50])
count = np.sum(np.isnan(a))
print(count)
```

```python
a = np.array([10, np.nan, np.nan, np.nan, np.nan, 40, 50])
count = np.sum(np.isnan(a))
print(count)
```

### `np.where()`

we can also find the position of Nan. Use `np.where()` together with `np.isnan()`

```python
a = np.array([10, np.nan, np.nan, np.nan, np.nan, 40, 50])
positions = np.where(np.isnan(a))
print(positions)
```

```python
a = np.array([
    [10, np.nan, np.nan],
    [np.nan, np.nan, 40]
])
count = np.sum(np.isnan(a))
print(count)
```

# Ways to handle NaN Values

There are three common ways to handle NaN values.
- remove NaN  : We remove NaN by using boolean indexing.
- ignore Nan during Calculation
- replace NaN

```python
# remove NaN
a = np.array([10, 20, 30, np.nan, 50])
print(a[~np.isnan(a)])
```

```python
a = np.array([
    [10, 20, np.nan],
    [40, np.nan, 60]
])

print(a[~np.isnan(a)])
```

```python
# ignore NaN during calculations
a = np.array([10, 20, 30, np.nan, 50])
print(np.nanmean(a))
```

```python
a = np.array([10, 20, 30, np.nan, 50])
print(np.nanmedian(a))
```

```python
a = np.array([10, 20, 30, np.nan, 50])
print(np.nanvar(a))
```

```python
# replace nan: we can replace nan according to the situation(mean according to dataset)
a = np.array([10, 20, 30, np.nan, 50])
print(np.nan_to_num(a, nan=0))
```

```python
a = np.array([10, 20, 30, np.nan, 50])
print(np.nan_to_num(a, nan=100))
```

```python
a = np.array([
    [10, 20, np.nan],
    [40, np.nan, np.nan]
])

print(np.nan_to_num(a, nan=100))
```

```python
a = np.array([
    [10, 20, np.nan],
    [40, np.nan, np.nan]
])

print(np.nan_to_num(a, nan=0))
```
