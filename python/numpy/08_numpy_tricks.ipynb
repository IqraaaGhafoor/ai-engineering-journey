# np.histogram()

Used to create frequency distribution. It divides data into intervals, called bins and counts how many values fall into each bin.

Used for:
- Data visualization
- Distribution analysis
- Exploratory data analysis
- Feature understanding

```python
a = np.array([10, 20, 30, 40, 50])
np.histogram(a)
```

In this example we have total 5 values. min value is 10 and max value is 50. So then range is 40.

`range = maximum - minimum`

here by default total bins are 10. So width or size of each bin `40 / 10 = 4` 

size of each bin = 4

we will start from the minimum value that is 10. 

`10 - 14 will be 1.`
`14 - 18 will be 0.`

and so on...



Returns two values: 

- hist: contains the number of data values present in each bin
- bins: contains the boundary values means it tells us where each group starts and ends.


```python
b = np.array([
    [10, 15, 20],
    [25, 30, 35],
    [40, 45, 50]
])

hist, bins = np.histogram(b)
print(hist)
print(bins)
```

# np.corrcoef()

Used to calculate correlation between variables. correlation means relation.

Used in:
- Feature selection
- Data analysis
- Machine Learning
- Feature selection
- Exploratory Data Analysis (EDA)

```python
hours = np.array([1, 2, 3, 4, 5])
marks = np.array([50, 60, 70, 80, 90])
np.corrcoef(hours, marks)
```

```python
a = np.array([
    [1, 2, 3, 4, 5],
    [50, 60, 70, 80, 90]
])

np.corrcoef(a)
```
