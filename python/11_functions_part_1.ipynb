### Creating a Function

- def starts a function definition.
- Function name identifies the function.
- () contains parameters.
- : starts the function body.
- Indentation defines which code belongs to the function.
- Defining a function does not run it; we have to call it.

```python
def greet():
    print("Hello")
    print("Welcome")

greet()
```

```python
def student():
    print("Name: Alex")
    print("Age: 28")
    print("Grade: A")

student()
```

```python
def fruits():
    print("Favorite: Mango")
    print("Not Favorite: Apple")
    
fruits()
```

### Calling a Function

- A function is executed by writing its name followed by ().
- def creates the function and function_name() calls it.
- Same function can be called many times.

```python
def welcome():
    print("Welcome!")

welcome()
welcome()
```

```python
def show_menu():
    print("1. Noddles")
    print("2. Chinese Rics")
    print("3. Soup")

show_menu()
```

### Parameters vs Arguments

- **Parameter** : variable written inside the function definition.
- **Argument** : actual value passed when calling the function.

```python
def greet(name):      #name is parametet
    print("Hello", name)

greet("Alex")   #Alex and other names are arguments
greet("Sara")
greet("Siri")
greet("Asbah")
```

```python
def add(a, b):
    print(a + b)

add(10, 20)
add(50, 30)
```

```python
def division(a, b):
    print(a / b)

division(10, 20)
division(50, 30)
```

```python
def power(x, y):
    print(x ** y)

power(10, 20)
power(50, 30)
```

```python
def subtract(x, y):
    print(x - y)

subtract(10, 20)
subtract(50, 30)
```

```python
def student(name, age, grade):
    print("Name:", name)
    print("Age:", age)
    print("Grade:", grade)

student("Alex", 28, "A")
```

### return

- return sends a value back from a function.
- The returned value can be stored in a variable.
- Returned values can be used in other calculations.
- return also stops the function at that point.

```python
def add(x, y):
    return x + y

result = add(10, 20)
print(result)
```

```python
def square(number):
    return number * number

result = square(7)
print(result)
```

```python
def multiply(x, y):
    return x * y

answer = multiply(6, 7)
print(answer)
```

```python
# print vs return
def add_print(x, y):
    print(x + y)

add_print(10, 20)
```

```python
def add_return(a, b):
    return a + b

result = add_return(100, 200)

print(result)
print(result * 2)
print(result + 100)
```

If a function does not have a return, Python returns None.

```python
def greet():
    print("Hello")

result = greet()
print(result)
```

```python
def test():
    return

result = test()
print(result)
```

Code after return in the same function is not executed.

```python
def test():
    print("Start")
    return 10
    print("End")

result = test()
print(result)
```

```python
def check():
    return "Done"
    print("This will not run")

print(check())
```

### Returning Multiple Values

- A function can return multiple values.
- Python automatically packs them into a tuple.
- This connects with Tuple packing and unpacking.

```python
def calculate(a, b):
    return a + b, a - b

result = calculate(250, 170)
print(result)
```

```python
def get_student():
    return "John", 20, "A"

name, age, grade = get_student()

print(name)
print(age)
print(grade)
```

### Positional Arguments

- Arguments are matched according to their position.
- First argument : first parameter.
- Second argument : second parameter.
- Order matters.

```python
def student(name, age):
    print("Name:", name)
    print("Age:", age)

student("Ali", 20)
```

```python
def numbers(a, b):
    print("a =", a)
    print("b =", b)

numbers(100, 200)
```

### Keyword Arguments

- Keyword arguments use the parameter name.
- Order does not matter when using keywords.
- They make function calls easier to understand.

```python
def student(name, age):
    print("Name:", name)
    print("Age:", age)

student(name="Alex", age=27)
```

```python
def introduce(name, age, city):
    print(name, age, city)

introduce(city="Islamabad", name="Anaya", age=22)
```

```python
# position and keyword argument

def student(name, age, city):
    print(name, age, city)

student("Ali", age=20, city="Karachi")
```

### Default Parameter Values

- A parameter can have a default value.
- If no argument is provided, the default is used.
- If an argument is provided, it replaces the default.

```python
def greet(name="Guest"):
    print("Hello", name)

greet()
greet("Alex")
greet("Siri")
```

```python
def power(number, exponent=2):
    return number ** exponent

print(power(5))
print(power(5, 3))
```

```python
def order_food(food="Pizza"):
    print("You ordered:", food)

order_food()
order_food("Burger")
order_food("Biryani")
```

### Docstrings

- A docstring describes what a function does.
- It is written immediately below the def line.
- Triple quotes """ """ are commonly used.
- Docstrings make code easier to understand.

```python
def greet():
    """Display a greeting message."""
    print("Hello")

greet()
```

```python
def square(number):
    """Return the square of a number."""
    return number ** 2

print(square(5))
```

```python
def add(x, y):
    """Return the sum of two numbers."""
    return x + y

print(add(70, 50))
```

```python
# function.__doc__ returns the function's docstring.

def greet():
    """Display a greeting message."""
    print("Hello")

print(greet.__doc__)
```

### Calling Functions from Other Functions

- One function can call another function.
- This helps divide a large task into smaller tasks.
- The called function can return a value to the calling function.

```python
def greet():
    print("Hello")

def welcome():
    greet()
    print("Welcome to Python")

welcome()
```

```python
def add(a, b):
    return a + b

def double_sum(a, b):
    result = add(a, b)
    return result * 2

print(double_sum(11, 9))
```

```python
def square(number):
    return number ** 2

def calculate(number):
    result = square(number)
    return result + 10

print(calculate(5))
```
