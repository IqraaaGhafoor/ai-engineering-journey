## `__init__()`

- `__init__()` is a special method.
- It runs automatically when an object is created.
- It is mainly used to initialize an object's data.
- It is often called the constructor in beginner-level explanations.

When we write:
`Student("Ali", 20)`

Python automatically runs `__init__()`.

```python
class Student:

    def __init__(self):
        print("Student object created")

student1 = Student()
student2 = Student()
```

## `__init__()` With Parameters

- Parameters allow us to give each object different data.
- Values are passed when creating the object.
- `self.name` store the data inside the object.

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age


student1 = Student("Ali", 20)
student2 = Student("Sara", 21)

print(student1.name)
print(student1.age)

print(student2.name)
print(student2.age)
```

## What is self?

- self refers to the **current object**.
- It allows each object to store its own data.
- `self.name` means the `name` belonging to the current object.
- self is normally the first parameter of an instance method.

```python
class Student:

    def __init__(self, name):
        self.name = name

student1 = Student("Ali")
student2 = Student("Sara")

print(student1.name)
print(student2.name)
```

```python
class fruit:
    
    def __init__(self, name):
        self.name = name

fruit1 = fruit("Mango")
fruit2 = fruit("Cherry")

print(fruit1.name)
print(fruit2.name)
```

## Understanding `self.name = name`

There are two different `name`s:

- `name` : parameter/local value
- `self.name` : attribute stored inside the object

`self.name = name` means:

Store the received value inside the current object.

```python
class Student:

    def __init__(self, name):
        self.name = name

student1 = Student("Ali")
print(student1.name)
```

## Instance Attributes

- Instance attributes belong to a specific object.
- They are commonly created using `self`.
- Different objects can have different values.

```python
class Student:

    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks


student1 = Student("Ali", 20, 85)
student2 = Student("Sara", 21, 92)

print(student1.name, student1.age, student1.marks)
print(student2.name, student2.age, student2.marks)
```

## Class Attributes

- A class attribute belongs to the class.
- It can be accessed by all objects.
- It is useful for information shared by objects.

```python
class Student:

    school = "Business School"

    def __init__(self, name):
        self.name = name


student1 = Student("John")
student2 = Student("Sara")

print(student1.name)
print(student2.name)

print(student1.school)
print(student2.school)
```

## Instance vs Class Attributes

### Instance Attribute
- Specific to one object.
- Usually created with `self`.
- Can have different values.

### Class Attribute
- Belongs to the class.
- Usually shared by objects.
- Used for common information.

```python
class Student:

    school = "Business School"   #classs attribute

    def __init__(self, name, age):
        self.name = name  #instance attribute
        self.age = age   #instance attribute

student1 = Student("John", 24)
student2 = Student("Sara", 26)

print(student1.name, student1.age)
print(student2.name, student2.age)

print(student1.school)
print(student2.school)
```

## Methods

- A function inside a class is called a method.
- Methods define the behavior of an object.
- Instance methods normally have `self` as the first parameter.
- Call a method using:
  `object.method()`

```python
class Student:

    def introduce(self):
        print("I am a student")


student1 = Student()
student1.introduce()
```

## Methods Using Instance Attributes

- Methods can access object data using `self`.
- `self.name` accesses the current object's `name`.
- Each object uses its own attributes.

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"My name is {self.name}")
        print(f"I am {self.age} years old")

student1 = Student("Ali", 20)
student2 = Student("Sara", 21)

student1.introduce()
student2.introduce()
```

- A class can contain many methods.
- Each method can perform a different task.
- Methods can use the object's attributes.

```python
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def introduce(self):
        print(f"My name is {self.name}")

    def show_marks(self):
        print(f"My marks are {self.marks}")


student1 = Student("John", 95)

student1.introduce()
student1.show_marks()
```

### Calling One Method From Another

- A method can call another method using `self`.
- Syntax:
  `self.method_name()`
- `self` refers to the current object.

```python
class Student:

    def introduce(self):
        print("I am a student")

    def welcome(self):
        self.introduce()
        print("Welcome to the class!")


student1 = Student()
student1.welcome()
```

### Method Calling With Object Data

When we call:

`student1.welcome()`

Inside the method:

`self` refers to `student1`.

Therefore: `self.introduce()` means `student1.introduce()`

```python
class Student:

    def __init__(self, name):
        self.name = name

    def introduce(self):
        print(f"My name is {self.name}")

    def welcome(self):
        self.introduce()
        print("Welcome!")


student1 = Student("Ali")

student1.welcome()
```

```python
class Student:

    school = "Business School"

    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks

    def introduce(self):
        print(f"My name is {self.name}")
        print(f"I am {self.age} years old")

    def show_marks(self):
        print(f"My marks are {self.marks}")

    def show_info(self):
        self.introduce()
        self.show_marks()


student1 = Student("John", 22, 95)
student1.show_info()
print(student1.school)
```
