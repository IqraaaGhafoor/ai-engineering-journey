## Arithmetic Operators

Arithmetic operators are used to perform calculations.

| Operator | Meaning | Example |
|---|---|---|
| `+` | Addition | `9 + 2` |
| `-` | Subtraction | `9 - 2` |
| `*` | Multiplication | `9 * 2` |
| `/` | Normal/True division | `9 / 2` |
| `//` | Floor division | `9 // 2` |
| `%` | Remainder | `9 % 2` |
| `**` | Power | `9 ** 2` |

```python
a = 13
b = 7

print(a + b)   # addition
print(a - b)   # subtraction
print(a * b)   # multiplication
print(a / b)   # division
print(a // b)  # floor division means without decimal points
print(a % b)   # remainder
print(a ** b)  # power
```

```python
number = 97

print(number + 197)
print(number / 65)
print(number // 65)
print(number ** 17)
```

```python
number = -67

print(number - 100)
print(number + 198)
print(number ** 14)
print(number // 55)  # not -1, as floors down towards negative infinity
print(number / 45)
```

## Comparison Operators

Compare values and return True or False.

- == → Equal
- != → Not equal, > → Greater than, < → Less than, >= → Greater than or equal
- <= → Less than or equal

= → Assignment

== → Comparison

Python allows multiple comparisons in one expression.

e.g: 18 <= age < 65

means: age >= 18 and age < 65

```python
x = 90
y = 55

print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
```

```python
age = 30

print(age >= 28)
print(age == 30)
print(age < 29)
```

## Logical Operators

Logical operators combine conditions.

- and → Both conditions must be True
- or → At least one condition must be True
- not → Reverses True/False

E.g:

age >= 18 and has_id

has_cash or has_card

not has_id

```python
age = 25
has_id = True

print(age >= 18 and has_id)
print(age >= 18 or has_id)
print(not has_id)
```

```python
has_cash = False
has_card = True

if has_cash or has_card:
    print("Payment possible")
else:
    print("Payment not possible")
```

Python can stop evaluating a logical expression when the result is already known.

False and something  → result is already False

True or something  → result is already True

Put safer conditions first when the second condition could cause an error.

```python
name = ""

if name and name[0] == "Z":
    print("Starts with Z")
else:
    print("No name or does not start with Z")
```

```python
name = "Zoraiz"

if name and name[0] == "Z":
    print("Starts with Z")
else:
    print("No name or does not start with Z")
```

##  Operator Precedence

Python use PEMDAS rule

p: parenthesis

E: exponent

M: Multiplication

D: Division

A: Addition

S: Subtraction

Python follows an order when evaluating expressions.

e.g 2 + 3 * 4 → 14

Multiplication happens before addition.

Use parentheses to make the order clear: (2 + 3) * 4 → 20

```python
print(12 + 13 * 24)
print((72 + 63) * 44)
```

## Control Flow Statements

Conditionals allow Python to make decisions.

### if
Runs when a condition is True.

### elif
Checks another condition.

### else
Runs when none of the previous conditions are True.

Python checks conditions from top to bottom and stops at the first True branch.

```python
score = int(input("Enter score: "))

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print(f"Grade: {grade}")
```

```python
age = 20
if age >= 18:
    print("you are eligible to vote.")
elif age >= 15:
    print("you are a toddler.")
elif age <= 6:
    print("you should go to school")
else:
    print("you are too young to vote.") 
```

```python
# calculator

num1 = float(input("Enter first number: "))
operator = input("Enter operator (+, -, *, /, //, %, **): ")
num2 = float(input("Enter second number: "))

if operator == "+":
    result = num1 + num2

elif operator == "-":
    result = num1 - num2

elif operator == "*":
    result = num1 * num2

elif operator == "/":
    if num2 == 0:
        print("Cannot divide by zero.")
    else:
        result = num1 / num2

elif operator == "//":
    if num2 == 0:
        print("Cannot divide by zero.")
    else:
        result = num1 // num2

elif operator == "%":
    if num2 == 0:
        print("Cannot divide by zero.")
    else:
        result = num1 % num2

elif operator == "**":
    result = num1 ** num2

else:
    print("Invalid operator.")

if operator in ["+", "-", "*", "**"]:
    print(f"Result: {result}")

elif operator in ["/", "//", "%"] and num2 != 0:
    print(f"Result: {result}")
```
