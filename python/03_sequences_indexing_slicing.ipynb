## Sequences

A **sequence** is any type where items are stored in a defined order and each
item has a position number (an index).

Ordered collection of items where each item can be acessed by its position (index) in the sequence.

All sequences support: indexing · slicing · len() · iteration



|  Type  | Example   | Mutable |
|--------|-----------|---------|
|  str   | "hello"   |  NO     |
|  list  | [1, 2, 3] |  Yes    |
|  tuple | (1, 2, 3) |  NO     |
|  range | range(5)  |  NO     |

Sets and dicts are **not** sequences — they can't be indexed by position.

```python
text = "Python"
numbers = [10, 20, 30]
colors = ("red", "green", "blue")
values = range(5)

print(text)
print(numbers)
print(colors)
print(values)
```

### String as a Sequence

A string is a sequence of characters.

Example: "Python"

The characters have indexes:

P y t h o n
0 1 2 3 4 5

Because strings are sequences, so we can use:

- indexing
- slicing
- len()
- loops
- membership testing

```python
word = "Python"

print(word[0])
print(word[3])
print(word[2])
print(len(word))
print("P" in word)
```

### List as a Sequence

A list is an ordered and mutable sequence.

Lists can contain different types of values.

```python
numbers = [10, 20, 30, 40]

print(numbers[0])
print(numbers[2])
```

```python
data = [10, "Python", 3.14, True]
print(data)
print(data[1])
```

### Tuple as a Sequence

A tuple is an ordered and immutable sequence.

Example: ("red", "green", "blue")

```python
colors = ("red", "green", "blue")

print(colors[0])
print(colors[-1])
print(len(colors))
print("green" in colors)
```

### Range as a Sequence

range() represents a sequence of integers.

range(5) represents: 0, 1, 2, 3, 4

The stop value is not included.

range() is useful when working with loops.

A range is not a list. Python does not need to create every number as a list in memory.

```python
numbers = range(5)
print(list(range(5)))
```

```python
numbers = range(2, 10)
print(list(range(2,10)))
```

```python
numbers = range(5)

print(numbers)
print(list(numbers))

print(numbers[0])
print(numbers[-1])
print(len(numbers))
```

Sequences commonly support:

len(sequence)
→ counts the number of items

sequence[index]
→ gets one item

sequence[start:stop]
→ gets multiple items

item in sequence
→ checks whether an item exists

item not in sequence
→ checks whether an item does not exist

for item in sequence
→ loops through the sequence

```python
numbers = [90, 80, 60, 40]

print(len(numbers))

print(numbers[1])

print(numbers[1:3])

print(20 in numbers)

print(100 not in numbers)

for number in numbers:
    print(number)
```

## What is Indexing?

Indexing means accessing one specific item from a sequence.

Syntax: sequence[index]

Python uses zero-based indexing.

This means the first item has index 0.

```python
word = "Python"

print(word[0])
print(word[1])
print(word[2])
print(word[3])
print(word[4])
print(word[5])
```

```python
word = "Python"

print(word[-1])
print(word[-2])
print(word[-3])
print(word[-6])
```

```python
numbers = [10, 20, 30, 40]

print(len(numbers))
print(numbers[3])
print(numbers[-1])
```

```python
numbers_list = [10, 20, 30]
numbers_tuple = (10, 20, 30)
numbers_range = range(3)

print(numbers_list[0])
print(numbers_tuple[0])
print(numbers_range[0])
```

## Nested Indexing

A sequence can contain another sequence.

Example:

students = [
    ["Ali", 20],
    ["Sara", 21],
    ["Ahmed", 19]
]

students[0] gives:

["Ali", 20]

students[0][0] gives:

"Ali"

The first index selects the inner sequence.
The second index selects an item from that inner sequence.

```python
students = [
    ["Ali", 20],
    ["Sara", 21],
    ["Ahmed", 19]
]

print(students[0])
print(students[0][0])
print(students[1][0])
print(students[2][1])
```

### Multiple-Level Indexing

We can have several levels of nested sequences.

Python evaluates indexing from left to right.

Example: data[0][1][0]

Each index moves one level deeper into the nested structure.

```python
data = [
    [
        ["Python", "ML"],
        ["DL", "AI"]
    ]
]

print(data[0])
print(data[0][0])
print(data[0][0][0])
```

```python
data = [
    [
        ["Ali", 20],
        ["Sara", 21]
    ],
    [
        ["Ahmed", 19],
        ["Ayesha", 22]
    ]
]

print(data[1])
```

```python
print(data[1][1])
```

```python
print(data[1][1][0])
```

```python
print(data[0][1][0])
```

## Slicing

Slicing means taking multiple items from a sequence.

Syntax:

sequence[start:stop:step]

start → where to begin (included)
stop → where to stop (excluded)
step → how many positions to move each time

Example:

numbers[1:4]

includes indexes:

1, 2, 3

but not index 4.

```python
numbers = [10, 20, 30, 40, 50]

print(numbers[1:4])
print(numbers[2:])
print(numbers[:4])
```

```python
numbers = [0, 1, 2, 3, 4, 5]

print(numbers[0:6:2])
print(numbers[::2])
```

#### Different Step Values

step = 1
every item

step = 2
every second item

step = 3

every third item

The step controls the movement through the sequence.

```python
numbers = [0, 1, 2, 3, 4, 5, 6, 7, 8]

print(numbers[::1])
print(numbers[::2])
print(numbers[::3])
```

```python
numbers = [0, 1, 2, 3, 4, 5]

print(numbers[::-1])
```

Strings can also be reversed using slicing.

word[::-1]

This creates a reversed string.

Remember that strings are immutable, so the original string is not modified.

```python
word = "Python"

reversed_word = word[::-1]

print(word)
print(reversed_word)
```

```python
numbers = [0, 1, 2, 3, 4, 5]

print(numbers[5:1:-1])
```

#### Step Cannot Be Zero

A slice step cannot be 0.

This is invalid:

numbers[::0]

Python raises: ValueError

### Copying with [:]

For lists, [:] is a common way to make a shallow copy.

Example:

b = a[:]

This creates a new outer list.

It is different from:

b = a

because b = a creates an alias to the same list.

```python
a = [1, 2, 3]

b = a[:]

b[0] = 100

print(a)
print(b)
```

b = a

Python does not create a new list.

Both variables refer to the same list.

Therefore, changing the list through b also affects what we see through a.

```python
a = [1, 2, 3]

b = a

b[0] = 99

print(a)
print(b)
```

## Shallow Copy

A shallow copy creates a new outer container but does not recursively copy nested mutable objects.

For a simple list:

b = a[:] is usually enough to create an independent outer list.

However, with nested lists, inner lists can still be shared.

```python
a = [[1, 2], [3, 4]]

b = a[:]

b[0][0] = 99

print(a)
print(b)
```

## Deep Copy

For nested structures where completely independent copies are required, Python provides deepcopy() from the copy module.

deepcopy() recursively copies nested objects.

This is different from [:], which creates only a shallow copy.

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][0] = 99

print(a)
print(b)
```

## Mutable vs Immutable

Mutable means an object can be changed after it is created. In short means changeable 

Immutable means an object cannot be changed in place after it is created. Not changeable.

For the four sequences:

list is mutable.

str is immutable.

tuple is immutable.

range is immutable.

```python
# Mutable
numbers = [10, 20, 30]
numbers[0] = 100

print(numbers)

# Immutable 
word = "Python"
colors = ("red", "green", "blue")
numbers_range = range(5)

print(word)
print(colors)
print(numbers_range)
```

```python
numbers = [10, 20, 30]

numbers[0] = 100
numbers.append(40)  #adding single item at the end of the list

print(numbers)
```

```python
name = "python"
result = name.upper()
print(name)
print(result)

name = name.upper()
print(name)
```

```python
name = "hello"
name = name.upper()
print(name)
```

```python
name = "Ali"
name = "Sara"  # Reassignment means changing what a variable refers to.

print(name)

numbers = [10, 20, 30]
numbers[0] = 100
print(numbers)
```

```python
data = ([1, 2], 3)
data[0].append(4)
print(data)
```

## What is Type Casting?

Type casting means converting a value from one data type to another.

In Python, functions such as:

int()
float()
str()
bool()
list()
tuple()

are commonly used for type conversion.

```python
number = "87"
number_int = int(number)
print(number_int)
print(type(number_int))
```

```python
# str to list

word = "Python"
letters = list(word)
print(letters)
```

```python
# str to tuple

word = "Python"
letters = tuple(word)
print(letters)
```

```python
# list to tuple

numbers = [1, 2, 3]
result = tuple(numbers)
print(result)
print(type(result))
```

```python
# tuple to list 

numbers = (1, 2, 3)
result = list(numbers)
result.append(4)
print(result)
```

```python
# range to list or tuple

numbers = range(5)

print(list(numbers))
print(tuple(numbers))
```

split() is a string method that breaks a string into smaller pieces.

```python
text = "Python is easy"

words = text.split()

print(words)
print(type(words))
```

When split() is used without an argument, Python splits around whitespace.

It handles multiple spaces conveniently.

```python
text = "python is easy"
print(text.split())
```
