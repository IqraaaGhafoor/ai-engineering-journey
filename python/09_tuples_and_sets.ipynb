## Tuples

Ordered collection of values that cannot be changed after creation. usually created by using ().

- Supports indexing and slicing
- Allows duplicate values
- Immutable (cannot be changed)
- Can contain different data types


```python
numbers = (10, 20, 30, 40)

print(numbers)
print(type(numbers))
```

```python
numbers = (10, 20, 30)
student = ("Ali", 20, 3.5, True)

a = (5)  #not tuple
b = (5,)

print(numbers)
print(student)

print(type(a))
print(type(b))
```

```python
# indexing

numbers = (10, 20, 30, 40, 50)

print(numbers[0])
print(numbers[2])
print(numbers[-1])
print(numbers[-2])
```

```python
#slicing

numbers = (10, 20, 30, 40, 50)

print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])
print(numbers[::-1])
```

## Tuple Packing

Packing means putting multiple values into one tuple.

Python can automatically create a tuple when multiple values are separated by commas.

```python
student = "Ali", 20, "Computer Science"

print(student)
print(type(student))
```

```python
# Tuple unpacking
# Unpacking means taking values from a tuple and assigning them to separate variables.
# The number of variables should normally match the number of values.

student = ("Ali", 20, "Computer Science")

name, age, department = student

print(name)
print(age)
print(department)
```

- *rest in the Middle

The `*` operator can collect multiple remaining values into a list.

```python
numbers = (10, 20, 30, 40, 50)

first, *middle, last = numbers

print(first)
print(middle)
print(last)
```

```python
# swapping variables

x = 10
y = 20

x, y = y, x

print(x)
print(y)
```

```python
def calculate(a, b):
    return a + b, a - b


addition, subtraction = calculate(10, 5)

print(addition)
print(subtraction)
```

```python
def calculate(a, b):
    return a + b, a - b


addition, subtraction = calculate(100, 50)

print(addition)
print(subtraction)
```

## Tuple Methods

Tuples have fewer methods because they are immutable.

Important methods:

- count() : counts occurrences of a value
- index() : finds the first index of a value

```python
numbers = (10, 20, 10, 30, 10)

print(numbers.count(10))
print(numbers.index(30))
```

```python
# membership - in

numbers = (10, 20, 30, 40)

print(20 in numbers)
print(100 in numbers)
print(100 not in numbers)
```

# Sets

A set is a collection of unique values.

- Unordered
- No duplicate values
- No indexing
- Mutable
- Very useful for membership testing
- Supports mathematical set operations

```python
numbers = {1, 2, 3, 4}

print(numbers)
print(type(numbers))
```

```python
# remove duplicates, sets keeps only one value of each
numbers = {1, 2, 2, 3, 3, 3, 4}
print(numbers)
```

```python
# `{}` creates an empty dictionary, NOT an empty set.
#  we use () to create an empty set.

empty_dict = {}
empty_set = set()

print(type(empty_dict))
print(type(empty_set))
```

```python
# Use `add()` to add one element to a set.
numbers = {1, 2, 3}

numbers.add(4)
print(numbers)
```

## update()

Use update() to add multiple elements.

```python
numbers = {1, 2, 3}

numbers.update([4, 5, 6])
print(numbers)
```

```python
numbers = {1, 2}

numbers.add(3)
print(numbers)

numbers.update([4, 5, 6])
print(numbers)
```

```python
# remove() removes a specific value from a set.

numbers = {10, 20, 30}

numbers.remove(20)
print(numbers)
```

```python
# `discard()` removes a value if it exists. If not, nothing happens and no error is raised.

numbers = {10, 20, 30}

numbers.discard(20)
numbers.discard(100)
print(numbers)
```

```python
# pop() removes and returns an element from a set.

numbers = {10, 20, 30}

removed_value = numbers.pop()
print("Removed:", removed_value)
print("Remaining:", numbers)
```

```python
# clear() removes all elements from a set.

numbers = {1, 2, 3, 4}

numbers.clear()
print(numbers)
```

```python
# membership - in

students = {"Ali", "Sara", "Alex", "John"}

print("Sara" in students)
print("John" in students)
print("John" not in students)
```

## Set Union

Union means combining all values from two sets without duplicates.

Operator:

A | B

Method:

A.union(B)

```python
A = {1, 2, 3}
B = {3, 4, 5}

print(A | B)
print(A.union(B))
```

```python
A = {12, 27, 73}
B = {32, 54, 85}

print(A | B)
print(A.union(B))
```

## Set Intersection

Intersection means finding values that exist in BOTH sets.

Operator:

A & B

Method:

A.intersection(B)

```python
A = {1, 2, 3}
B = {3, 4, 5}

print(A & B)
print(A.intersection(B))
```

```python
A = {91, 72, 63}
B = {63, 74, 95}

print(A & B)
print(A.intersection(B))
```

```python
# difference 
A = {1, 2, 3}
B = {3, 4, 5}

print(A - B)
print(B - A)
```

### Symmetric Difference

Symmetric difference finds values that are in either set, but NOT in both.

Operator:

A ^ B

Method:

A.symmetric_difference(B)

```python
A = {1, 2, 3}
B = {3, 4, 5}

print(A ^ B)
print(A.symmetric_difference(B))
```

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print("Union:", A | B)
print("Intersection:", A & B)
print("A - B:", A - B)
print("B - A:", B - A)
print("Symmetric Difference:", A ^ B)
```

### issubset()

A set is a subset if every element in it exists in another set.

Syntax: A.issubset(B)

we can use this operator A <= B instead of this A.issubset(B)

```python
A = {1, 2}
B = {1, 2, 3, 4}

print(A.issubset(B))
```

```python
A = {1, 2}
B = {1, 2, 3, 4}

print(A <= B)
```

## issuperset()

A set is a superset if it contains every element of another set.

Syntax: B.issuperset(A)

we can use also this B >= A operator.

```python
A = {1, 2}
B = {1, 2, 3, 4}

print(B >= A)
```

```python
# converting a list to a set

numbers = [1, 2, 2, 3, 3, 4, 4]
unique_numbers = set(numbers)
print(unique_numbers)
```

```python
# looping through a set 

students = {"Alex", "John", "Anna", "Z.Tao"}

for student in students:
    print(student)
```
