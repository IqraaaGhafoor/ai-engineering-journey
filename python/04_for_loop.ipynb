### Range 

range() generates numbers on demand. 

range(5)          # 0, 1, 2, 3, 4        (stop only)

range(2, 6)       # 2, 3, 4, 5           (start, stop)

range(0, 20, 5)   # 0, 5, 10, 15         (start, stop, step)

range(10, 0, -1)  # 10, 9, 8 ... 1       (counting down)

stop value not included.

#### enumerate()

sometimes you need both the position and item.

```python
fruits = ["apple", "cherry", "mango"]

for i, fruit in enumerate(fruits):
    print(i, fruit)
```

```python
number = [20, 40, 60, 80, 10]

for i, numbers in enumerate(number):
    print(i, numbers)
```

```python
marks = [80, 87, 95, 100]

for i, mark in enumerate(marks):
    print(i, mark)
```

```python
subjects = ["Python", "Math", "AI"]

for number, subject in enumerate(subjects):
    print(number, subject)
```

```python
subjects = ["Python", "Math", "AI"]

for number, subject in enumerate(subjects, start=1):
    print(f"{number}. {subject}")
```

```python
name = ["John", "David"]
ages = [29, 38]

for names, name in enumerate(name):
    print(names, name)

for i, age in enumerate(ages):
    print(i, age)
```

```python
name = ["John", "David"]
ages = [29, 38]

for i, (person, age) in enumerate(zip(name, ages)):
    print(i, person, age)
```

### zip() 

zip() allows me to work with two or more lists at the same time.

For example, 

names = ["Ali", "Sara", "Zain"]

scores = [85, 92, 78]

I want:

Ali → 85
Sara → 92
Zain → 78

We can use zip() to create these pairs.

```python
names = ["Ali", "Sara", "Zain"]
scores = [85, 92, 78]

for name, score in zip(names, scores):
    print(name, score)
```

```python
# zip() stops at the shorter list.

a = [1, 2, 3, 4, 5]
b = ["x", "y"]

for number, letter in zip(a, b):
    print(number, letter)
```

### break

break means: **"Stop the loop completely."**

When Python reaches break, it immediately leaves the loop.

This is useful when we have already found what we were looking for.

```python
for i in range(1, 11):

    if i == 6:
        break

    print(i)
```

```python
names = ["Ali", "Sara", "John", "Omar"]

for name in names:

    if name == "John":
        print("John found!")
        break
```

### continue

continue means: **"Skip this item and move to the next one."**

Unlike break, it does NOT stop the whole loop.

like we can use it to skip even numbers.

```python
for i in range(1, 11):

    if i % 2 == 0:
        continue

    print(i)
```

#### break vs continue

##### break
Stops the entire loop.

##### continue
Skips only the current iteration and continues with the next item.

### for...else

A for loop can also have an else.

The else runs when the loop finishes normally without using break.

This is especially useful when searching for something.

In simple words: Search for this item. If you don't find it, run the else block.

```python
names = ["Ali", "Sara", "John"]

target = "Omar"

for name in names:

    if name == target:
        print("Found!")
        break

else:
    print("Not found!")
```

```python
names = ["Ali", "Sara", "David"]

target = "Sara"

for name in names:

    if name == target:
        print("Found!")
        break

else:
    print("Not found!")
```

### Accumulator

An **accumulator** is a variable that stores and updates a result during a loop.

### Important points:
- Initialize it **before** the loop.
- Update it **inside** the loop.
- Common uses: **sum, count, building a list, finding largest/smallest**.

```python
numbers = [100, 200, 80]

total = 0

for number in numbers:
    total += number

print(total)
```

```python
# building a list

numbers = [1, 2, 3, 4]

squares = []

for number in numbers:
    squares.append(number ** 2)

print("Squares:", squares)
```

```python
numbers = [5, 10, 15, 20]

count = 0

for number in numbers:
    if number > 10:
        count += 1

print("Numbers greater than 10:", count)
```

```python
# find the largest

numbers = [25, 10, 45, 30]

largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

print("Largest:", largest)
```
