## What is OOP?

- OOP = Object-Oriented Programming.
- It is a way of organizing code around objects.
- Objects represent things/entities in a program.
- Objects can contain data and behavior.
- Python heavily uses objects.

```python
name = "Alice"          # string object
numbers = [11, 22, 33]     # list object
age = 20                # integer object

print(type(name))
print(type(numbers))
print(type(age))
```

## What is a Class?

- A class is a **blueprint/template** for creating objects.
- `class` is the keyword used to create a class.
- A class defines what objects created from it can have/do.
- Think of a class as a design, not the actual object.

```python
class Students:
    pass
```

## Class vs Object

### Class
- Blueprint/template.
- Defines the structure.
- Used to create objects.

### Object
- Actual instance of the class.
- Created from the class.
- Each object can represent a different entity.

## What is an Object?

- An object is an **actual instance** of a class.
- We create an object by calling the class.
- One class can create many objects.

Example:

Student is class  
student1 is object  
student2 is object

```python
class Student:
    pass


student1 = Student()
student2 = Student()

print(type(student1))
print(type(student2))
```

## 5. Why Do Classes Exist?

Classes help us **bundle related data and functions together**.

Without classes:
- Code can become difficult to organize.
- Related information may be scattered.

With classes:
- Related data : attributes
- Related functions : methods
- Both can be organized around one object.

Main idea is:

Class = blueprint for objects

Object = data + behavior

```python
# Without a class

student1_name = "Ali"
student1_age = 20

student2_name = "Sara"
student2_age = 21

print(student1_name, student1_age)
print(student2_name, student2_age)
```

## Creating a Simple Class

Syntax:

    # class body

```python
class Student:
    pass


class dinner:
    pass


class BankAccount:
    pass

class fruit:
    pass
```

## Creating Multiple Objects

- A single class can create many objects.
- Each object is a separate instance.
- Objects are created by writing:

object_name = ClassName()

```python
class Student:
    pass


student1 = Student()
student2 = Student()
student3 = Student()

print(type(student1))
print(type(student2))
print(type(student3))
```

```python
class fruit:
    pass


fruit1 = fruit()
fruit2 = fruit()
fruit3 = fruit()

print(type(fruit1))
print(type(fruit2))
print(type(fruit3))
```
