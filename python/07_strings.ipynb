##  Strings

A **string** is text written inside quotes or a sequence of characters written inside quotes.

**Imp point:** Strings are immutable, so string methods return a new string instead of changing the original string.

Python has different ways to write strings:

| Example | Name | Use |
|---|---|---|
| `'Hello'` | Single quotes | Normal text/string |
| `"Hello"` | Double quotes | Normal text, especially when text has `'` |
| `'''Hello'''` or `"""Hello"""` | Triple quotes | Multi-line text |

- Triple quotes are useful for **multi-line strings**.

````python
name = "John"
city = "USA"
message = "Hello Python!"

print(name)
print(city)
print(message)
````

````python
name1 = 'Justin'
name2 = "Justin"

message1 = "I'm learning Python"
message2 = 'He said "Hello"'

print(name1)
print(name2)
print(message1)
print(message2)
````

````python
name = "Alex"
text = 'Hy'

print(name)
print(text)
````

### Common Mistakes

**1. Forgetting the quotation marks**

Without quotes, Python treats the text as a variable name and raises NameError.

```python
# Wrong
name = John

# Correct
name = "John"
```

**2. Mismatched quotation marks**

A string must open and close with the same quote character.

```python
# Wrong
text = "Hello'

# Correct
text = "Hello"
text = 'Hello'
```

**3. Triple quotes do not create a special data type**

Triple quotes are just another way to write a string, not a different type.

```python
type("Hello")      # <class 'str'>
type("""Hy""")  # <class 'str'>
```

They are useful because the string can span multiple lines and can contain
single or double quotes without escaping.

## Escape Sequences

An escape sequence starts with a backslash \.

Here are the mportant escape sequences:

- \n → new line
- \t → tab
- \\ → backslash
- \' → single quote
- \" → double quote

````python
print("Hello\nWorld")

print("Name:\tFluffy")

print("C:\\Users\\John")

print('I\'m learning Python')

print("He said \"Hello\"")
````

#### Common Mistake
**Backslashes have a special meaning**

A backslash starts an escape sequence, so it does not always appear as a
backslash in the output. Write \\ when you want an actual backslash.

```python
# Wrong
path = "C:\Users\siri"

# Correct
path = "C:\\Users\\siri"

# Also correct (raw string)
path = r"C:\Users\siri"
```

````python
path = "C:\\Users\\Alex\\notes"

print(path)
````

## Raw Strings

A raw string starts with r: r"text"

Raw strings treat backslashes mostly as normal characters.

They are especially useful for:
- Windows file paths
- Regular expressions

Example: r"C:\Users\notes"

**Raw strings do not ignore every rule**

Do not think `r""` means that Python ignores every rule. Raw strings mainly
prevent backslashes from being interpreted as escape sequences.

Also remember: a raw string cannot normally end with a single backslash.

```python
# Wrong
path = r"C:\Users\"

# Correct
path = "C:\\Users\\"
```

````python
path = r"C:\Users\notes"   # avoid path = r"C:\Users\"

print(path)
````

## Case Methods

Case methods change capitalization.

upper() → ALL UPPERCASE

lower() → all lowercase

title() → First Letter Of Each Word

capitalize() → First character uppercase

Imp:
String methods return a new string.
They do not change the original string.

````python
text = "hello, what's up?"

print(text.upper())
print(text.lower())
print(text.title())
print(text.capitalize())

print(text)
````

````python
text = "hello"

text = text.upper()

print(text)
````

## Whitespace Methods

Whitespace includes spaces, tabs, and new lines.

strip() → removes whitespace from both sides

lstrip() → removes whitespace from the left

rstrip() → removes whitespace from the right

These methods do NOT remove spaces between words.

````python
text = "   Python is easy   "

print("strip:", text.strip())
print("lstrip:", text.lstrip())
print("rstrip:", text.rstrip())
````

````python
text = "   Hello World   "

text = text.strip()

print(text)
````

## find()

find() searches for a substring.

If found: returns its index.

If not found: returns -1.

Python indexing starts at 0.

````python
text = "Python is easy"

print(text.find("Python"))
print(text.find("easy"))
print(text.find("Java"))
````

````python
text = "Python"

position = text.find("Java")

if position == -1:
    print("Not found")
else:
    print("Found at:", position)
````

````python
text = "anaconda"

position = text.find("python")

if position == -1:
    print("Not found")
else:
    print("Found at:", position)
````

## index()

index() also searches for a substring.

If found: returns the index.

If not found: raises ValueError.

Main difference:

find() → -1

index() → ValueError

````python
text = "Python is easy"

print(text.index("Python"))
````

Imp point:

find() = safer search because it returns -1

index() = raises an error when the item is missing

````python
text = "Python"

if text.find("Java") == -1:
    print("Java was not found")
````

## count()

count() tells us how many times a substring occurs.

````python
text = "banana"

print(text.count("a"))
print(text.count("n"))
print(text.count("z"))
````

````python
text = "banana"

print(text.count("a"))
print(text.count("n"))
print(text.count("z"))
print(text.count("d"))
print(text.count("b"))
````

**count() is not the same as len()**

count() counts how many times a substring appears; it does not count the
length of the string.

````python
text = "banana"

print("a count:", text.count("a"))
print("length:", len(text))
````

## startswith() and endswith()

startswith() checks the beginning of a string.

endswith() checks the ending of a string.

Both return True or False.

Useful for:
- URLs
- usernames
- file extensions
- validation

````python
website = "https://example.com"
filename = "report.pdf"

print(website.startswith("https"))
print(filename.endswith(".pdf"))
````

````python
# These methods are case sensitive

text = "Python"

print(text.startswith("P"))
print(text.startswith("p"))

print(text.endswith("n"))
print(text.endswith("N"))
````

## replace()

replace() replaces one piece of text with another.

Syntax: string.replace(old, new)

You can also limit the number of replacements:

string.replace(old, new, count)

````python
text = "I like C++"

new_text = text.replace("C++", "Python")

print(new_text)
print(text)
````

replace does not change the original string.

````python
text = "Java is easy"

text = text.replace("Java", "Python")

print(text)
````

````python
text = "cat cat cat"

print(text.replace("cat", "dog"))
print(text.replace("cat", "dog", 2))
````

## String Checking, Splitting & Joining

- `isalpha()` → checks if all characters are letters.
- `isdigit()` → checks if all characters are digits.
- `isalnum()` → checks if all characters are letters or digits.
- `isspace()` → checks if all characters are whitespace.
- These methods return **`True` or `False`**.

- `split()` → converts a **string → list**.
- `split(separator)` → splits the string wherever the separator appears.
- `join()` → converts a **list of strings → one string**.
- `" ".join(list)` → joins items using a space.

- **Common pattern:** `split → process → join`.

`join()` works with **strings only**.  
`input()` always returns a **string**.

````python
print("Python".isalpha())
print("12345".isdigit())
print("Python123".isalnum())
print("   ".isspace())

print("Python123".isalpha())
print("Python!".isalnum())
````

````python
print("Python".isalpha())
print("Python13".isalpha())
print("Python Code".isalpha())
print("Python!".isalpha())
````

````python
age = "30"

print(age.isdigit())
print(type(age))

if age.isdigit():
    print("Contains only digits")
````

````python
print("   ".isspace())
print("\t".isspace())
print("\n".isspace())

print("Hello".isspace())
print(" Hello ".isspace())
````

````python
sentence = "Python is very easy"

words = sentence.split()

print(words)
print(type(words))
````

````python
# split with a seperator

data = "apple,banana,mango"
print(data.split(","))

date = "2026-09-06"
print(date.split("-"))

path = "home/user/python"
print(path.split("/"))
````

````python
words = ["Python", "is", "easy"]

print(" ".join(words))
print(", ".join(words))
print("-".join(words))
print("".join(words))
````

````python
sentence = "python is very easy"

words = sentence.split()

for i in range(len(words)):
    words[i] = words[i].title()

result = " ".join(words)

print(result)
````

## f-Strings

f-strings allow variables and expressions to be placed directly inside a string.

Syntax: f"{variable}"

Example: f"{name} is {age}"

f-strings are usually cleaner than using + and str().

````python
name = "Alex"
age = 20

print(f"{name} is {age} years old.")
````

````python
price = 100
a = 10
b = 20
name = "john"

print(f"Double price: {price * 2}")
print(f"Sum: {a + b}")
print(f"Name: {name.upper()}")
````

### f-String Formatting

Useful format specifiers:

{value:.2f}
→ 2 digits after the decimal

{n:,}
→ comma separators for large numbers

````python
price = 123.456789
number = 1000000

print(f"{price:.2f}")
print(f"{number:,}")
print(f"${price:,.2f}")
````

## 24. Common Beginner Mistakes

**1. Forgetting quotation marks**

```python
# Wrong
name = John

# Correct
name = "John"
```

**2. Incorrect quotes**

An apostrophe inside single quotes ends the string early.

```python
# Wrong
text = 'I'm happy'

# Correct
text = "I'm happy"
text = 'I\'m happy'
```

**3. Forgetting that strings are immutable**

You cannot change a string in place; you build a new one.

```python
# Wrong
name[0] = "H"

# Correct
name = "H" + name[1:]
```

**4. Thinking `strip()` removes spaces between words**

It only removes whitespace from the two ends.

```python
"  hello world  ".strip()   # "hello world"
```

**5. Confusing `find()` and `index()`**

Both search for a substring, but they behave differently when it is missing.

```python
"banana".find("z")    # -1
"banana".index("z")   # ValueError
```

**6. Confusing `split()` and `join()`**

```python
"a b c".split()          # ['a', 'b', 'c']   string to list
" ".join(['a','b','c'])  # "a b c"           list to string
```

**7. Using `join()` incorrectly**

The separator is the string you call the method on.

```python
# Wrong
words.join(" ")

# Correct
" ".join(words)
```

**8. Trying to join integers directly**

`join()` needs strings, so convert first.

```python
# Wrong
" ".join([1, 2, 3])

# Correct
" ".join(str(n) for n in [1, 2, 3])
```

**9. Forgetting that `input()` returns a string**

```python
age = input("Age: ")     # "20", not 20
age = int(input("Age: "))
```

**10. Using `is` instead of `==` to compare string values**

`==` compares the text; `is` compares whether they are the same object.

```python
a = "hello world"
b = "hello world"
a == b    # True
a is b    # may be False
```

**11. Forgetting that indexing starts at 0**

```python
"Python"[0]   # "P"
"Python"[1]   # "y"
```

**12. Forgetting that string comparisons are case-sensitive**

```python
"Alice" == "alice"                  # False
"Alice".lower() == "alice".lower()  # True
```

````python
# Correct string creation
name = "Alex"

# Correct comparison
if name == "Alex":
    print("Match")

# Correct cleaning
name = name.strip()

# Correct case conversion
name = name.lower()

# Correct split
words = "Python is easy".split()

# Correct join
sentence = " ".join(words)

print(sentence)
````
