```python
import json
```

```python
# creating student class

class Student:
    def __init__(self, student_id, name, age, course):
        self.student_id = student_id
        self.name = name
        self.age = age
        self.course = course

    def __str__(self):
        return f"{self.student_id} | {self.name} | {self.age} | {self.course}"
```

```python
# students dictionary

students = {}
```

```python
# adding student 
def add_student():
    student_id = input("Enter student ID: ").strip()

    if not student_id:
        print("Student ID cannot be empty.")
        return

    if student_id in students:
        print("Student ID already exists.")
        return

    name = input("Enter student name: ").strip()

    if not name:
        print("Name cannot be empty.")
        return

    try:
        age = int(input("Enter student age: "))

        if age <= 0:
            print("Age must be greater than 0.")
            return

    except ValueError:
        print("Age must be a valid number.")
        return

    course = input("Enter course: ").strip()

    if not course:
        print("Course cannot be empty.")
        return

    student = Student(student_id, name, age, course)
    students[student_id] = student

    print("Student added successfully.")
```

```python
# view student
def view_students():
    if not students:
        print("No students found.")
        return

    print("\nStudent Records")
    print("-" * 45)

    for student in students.values():
        print(student)
```

```python
# search student 
def search_student():
    student_id = input("Enter student ID to search: ").strip()

    if student_id in students:
        print("\nStudent Found:")
        print(students[student_id])
    else:
        print("Student not found.")
```

```python
# update student 

def update_student():
    student_id = input("Enter student ID to update: ").strip()

    if student_id not in students:
        print("Student not found.")
        return

    student = students[student_id]

    print("\nLeave a field empty to keep the current value.")

    name = input(f"Enter new name [{student.name}]: ").strip()

    if name:
        student.name = name

    age_input = input(f"Enter new age [{student.age}]: ").strip()

    if age_input:
        try:
            age = int(age_input)

            if age <= 0:
                print("Age must be greater than 0.")
                return

            student.age = age

        except ValueError:
            print("Age must be a valid number.")
            return

    course = input(f"Enter new course [{student.course}]: ").strip()

    if course:
        student.course = course

    print("Student updated successfully.")
```

```python
# delete student 
def delete_student():
    student_id = input("Enter student ID to delete: ").strip()

    if student_id not in students:
        print("Student not found.")
        return

    print(f"Student: {students[student_id]}")

    confirm = input("Are you sure you want to delete this student? (y/n): ").lower()

    if confirm == "y":
        del students[student_id]
        print("Student deleted successfully.")
    else:
        print("Deletion cancelled.")
```

```python
# save studnet 
def save_students():
    data = {}

    for student_id, student in students.items():
        data[student_id] = {
            "student_id": student.student_id,
            "name": student.name,
            "age": student.age,
            "course": student.course
        }

    try:
        with open("students.json", "w") as file:
            json.dump(data, file, indent=4)

        print("Students saved successfully.")

    except OSError as e:
        print(f"Error saving students: {e}")
```

```python
def load_students():
    try:
        with open("students.json", "r") as file:
            data = json.load(file)

        for student_id, info in data.items():
            students[student_id] = Student(
                info["student_id"],
                info["name"],
                info["age"],
                info["course"]
            )

        print("Students loaded successfully.")

    except FileNotFoundError:
        print("No saved student file found. Starting with an empty record.")

    except json.JSONDecodeError:
        print("The student file contains invalid JSON.")
```

```python
def menu():
    while True:
        print("\n" + "=" * 40)
        print("       STUDENT RECORD SYSTEM")
        print("=" * 40)
        print("1. Add Student")
        print("2. View Students")
        print("3. Search Student")
        print("4. Update Student")
        print("5. Delete Student")
        print("6. Save Students")
        print("7. Exit")
        print("=" * 40)

        choice = input("Enter your choice: ").strip()

        if choice == "1":
            add_student()

        elif choice == "2":
            view_students()

        elif choice == "3":
            search_student()

        elif choice == "4":
            update_student()
            break

        else:
            print("Invalid choice. Please select 1-4.")
```

```python
load_students()
menu()
```
